```python
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv("data/processed/cleaned_data.csv", parse_dates=["transaction_date"])
```

```python
df['year'] = df['transaction_date'].dt.year
df['month'] = df['transaction_date'].dt.month
df['quarter'] = df['transaction_date'].dt.quarter
df['day_of_week'] = df['transaction_date'].dt.dayofweek
```

```python
df['revenue_per_unit'] = df['total_spent'] / df['quantity']
```

```python
df['is_bulk'] = (df['quantity'] >= 5).astype(int)
```

```python
def segment_spent(x):
    if x < 60:
        return 'Low'
    elif x < 180:
        return 'Medium'
    else:
        return 'High'

df['spent_segment'] = df['total_spent'].apply(segment_spent)
```

```python
df['discount_flag'] = df['discount_applied'].astype(int)
```

```python
df['category_code'] = df['category'].astype('category').cat.codes
```

```python
df['location_code'] = df['location'].map({'Online': 0, 'In-store': 1})
```

```python
payment_map = {'Cash': 0, 'Credit Card': 1, 'Digital Wallet': 2}
df['payment_code'] = df['payment_method'].map(payment_map)
```
